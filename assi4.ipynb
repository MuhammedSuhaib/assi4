## List Methods Practice

## 1. Create and Modify Lists

```python
fruits = ["apple", "banana", "orange"]
fruits.append('grape')
fruits.insert(1,'mango')
print('fruits: ', fruits)
```

## 2. List Operations

```python
numbers=[ 10, 20, 30, 40, 50]
more_numbers=[60, 70, 80]
all_numbers= numbers + more_numbers
numbers_copy = all_numbers.copy()
numbers_copy.reverse()
print('all_numbers: ', all_numbers)
print('numbers_copy: ', numbers_copy)
```

## 3. Sorting and Counting

```python
scores =[85, 92, 78, 65, 92, 85, 74]
scores.sort()
print('ascending: ', scores)
scores.sort(reverse=True)
print('descending: ', scores)
print( scores.count(92))
print('Highest:', max(scores))
print('Lowest:', min(scores))
```

## 4. List Manipulation

```python
letters=['a', 'b', 'c', 'd', 'e']
print('letters: ', letters)
print('start_letters: ', letters[0:3])
print('end_letters: ',  letters[-2:])
print('index of c: ', letters.index("c"))
letters[3] = 'z'
print('letters: ',letters )
```

## Dictionary Methods Practice

## 1. Create and Access Dictionaries

```python

student={
"name": "John",
"age": 20,
"grade": "A",
"courses": ["Math", "Science", "History"]
}
print('name: ', student["name"])
print('courses: ', student["courses"])

student["email"]= "john@example.com"
print('student: ', student)
```

## 2. Modify Dictionary Values

```python
student["age"]=21
student["courses"].append("Computer Science")
student["grade"]="A+"
print('student: ', student)
```

## 3. Dictionary Operations

```python
student_address={
"street": "123 College Ave",
"city": "New York",
"zip": "10001"
}
# combine dictionaries
student_info= {**student, **student_address}
# remove zip
student_info.pop('zip')
# chech if number exists
print(student_info.get("phone", "phone does not exist"))
print("Final:", student_info)
```

## 4. Dictionary Methods

```python
word_count = {
"hello": 5,
"world": 10,
"python": 15
}
copy_dic =word_count.copy()
print('word_count.keys(): ', word_count.keys())
print('word_count.values(): ', word_count.values())
print('word_count',word_count)
word_count.clear()
print('copy_dic: ', copy_dic)
print('Empty: ', word_count)
```
