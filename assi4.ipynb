## List Methods Practice

## 1. Create and Modify Lists

```python
fruits = ["apple", "banana", "orange"]
fruits.append('grape')
fruits.insert(1,'mango')
print('fruits: ', fruits)
```

## 2. List Operations

```python
numbers=[ 10, 20, 30, 40, 50]
more_numbers=[60, 70, 80]
all_numbers= numbers + more_numbers
numbers_copy = all_numbers.copy()
numbers_copy.reverse()
print('all_numbers: ', all_numbers)
print('numbers_copy: ', numbers_copy)
```

## 3. Sorting and Counting

```python
scores =[85, 92, 78, 65, 92, 85, 74]
scores.sort()
print('ascending: ', scores)
scores.sort(reverse=True)
print('descending: ', scores)
print('92 appears', scores.count(92),'times')
print('Highest:', max(scores))
print('Lowest:', min(scores))
```

## Shopping Discount Calculator

```python
purchase_amount: int = int(input("Enter the purchase amount: "))
is_loyalty_member_input = input("Is the customer a loyalty member? (True/False): ")
is_loyalty_member: bool = is_loyalty_member_input.lower() == 'true'

day_of_week: str = input("Enter the day of the week: ")
is_holiday_season_input = input("Is it a holiday season? (True/False): ")
is_holiday_season: bool = is_holiday_season_input.lower() == 'true'

discount = 0
dis = {}

if purchase_amount <= 0:
    print("Invalid purchase amount.")
else:
    if is_loyalty_member:
        discount += purchase_amount * 0.05
        dis['is_loyalty_member'] = "5%"
    if purchase_amount > 100:
        discount += purchase_amount * 0.10
        dis['100+ purchase'] = "10%"
    elif purchase_amount > 50:
        discount += purchase_amount * 0.05
        dis['50+ purchase'] = "5%"
    if is_holiday_season:
        discount += purchase_amount * 0.08
        dis['is_holiday_season'] = "8%"
    if day_of_week.lower() in ['saturday', 'sunday']:
        discount += purchase_amount * 0.02
        dis['weekend'] = "2%"

    final_amount = purchase_amount - discount

    print("Original purchase amount: $", purchase_amount)
    print("Discount Type:", dis)
    print("Total Discount: $", round(discount, 2))
    print("Final Amount: $", round(final_amount, 2))
```

## While Loop Assignment - Interactive Game

```python
sec_num = 33
guess_count = 0

while True:
    resp = int(input("Guess the number (1-50): "))
    guess_count += 1
    if resp == sec_num :
        print('correct')
        break
    if resp in range(1,30):
        print('too low')
    if resp in range(36,49):
        print('too high')
    if resp in range(30,36):
        print("near")

print("You guessed it in", guess_count, "tries.")
```

## Student Grade Management System

```python
students = [
    {
        'name': 'Rehman',
        'id': 1,
        'grades': {'web3.0': 85, 'metaverse': 90, 'Agentic AI': 88}
    },
    {
        'name': 'Faraz',
        'id': 2,
        'grades': {'web3.0': 78, 'metaverse': 82, 'Agentic AI': 80}
    },
    {
        'name': 'Farhan',
        'id': 3,
        'grades': {'web3.0': 92, 'metaverse': 88, 'Agentic AI': 91}
    }
]

# Function to calculate average
def calculate_avg(grades):
    return sum(grades.values()) / len(grades)

# Display all students and their grades 
highest_avg = 0
top_student = None
print("Student Info:\n")

for student in students:
    print("Name:", student['name'])
    print("Grades:", student['grades'])
    
    avg = calculate_avg(student['grades'])
    print("Average Grade:", round(avg, 2), "\n")

    if avg > highest_avg:
        highest_avg = avg
        top_student = student

print('='*50)

print("Top Student:", top_student['name'], "with average", round(highest_avg, 2))

print('='*50)

# Students with 80+ average
print("\nStudents with 80+:")
for s in students:
    avg = calculate_avg(s['grades'])
    if avg > 80:
        print(s['name'], round(avg, 2))

print("\nStudents with 80+ grades in web3.0")
ttl_stds = len(students)
total = 0

for std in students:
    total += std['grades']['web3.0']
print('total: ', total)

avrg = total / ttl_stds
print('avrg:', avrg)

# Adding a new student
new_student = {
    'name': 'Hassan',
    'id': 4,
    'grades': {'web3.0': 87, 'metaverse': 89, 'Agentic AI': 84}
}
students.append(new_student)
print("\nAdded new student:", new_student['name'])
```
